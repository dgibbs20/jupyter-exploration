```python
# Install PyGame Learning Environment (PLE) from GitHub
!git clone https://github.com/ntasfi/PyGame-Learning-Environment.git  # Clone the repo
%cd PyGame-Learning-Environment  # Move into the cloned directory
!pip install -e .  # Install the environment in editable mode for real-time code updates

# Import necessary libraries
import numpy as np  # For numerical operations
import random  # For random action selection and sampling
import cv2  # For image processing (grayscale and resizing)
import tensorflow as tf  # For building and training the neural network
from tensorflow.keras.models import Model  # Base model class
from tensorflow.keras.layers import Dense, Flatten, Conv2D, Input  # Neural network layers
from collections import deque  # Efficient memory buffer for experience replay
from ple.games.flappybird import FlappyBird  # Import Flappy Bird game logic
from ple import PLE  # Game environment wrapper

# Initialize the Flappy Bird game inside PLE
game = FlappyBird()  # Create the game instance
env = PLE(game, fps=30, display_screen=False)  # Wrap the game with PLE engine
env.init()  # Start the game environment
actions = env.getActionSet()  # Get available actions (flap or do nothing)

# Preprocessing function for game frames
def preprocess_frame(frame):
    frame = cv2.cvtColor(frame, cv2.COLOR_RGB2GRAY)  # Convert RGB to grayscale
    frame = cv2.resize(frame, (84, 84))  # Resize frame to 84x84 pixels
    frame = frame / 255.0  # Normalize pixel values to range [0, 1]
    return frame.reshape((84, 84, 1))  # Add channel dimension for CNN input

# Define the Deep Q-Learning Agent class
class DQNAgent:
    def __init__(self, state_shape, action_size):
        self.state_shape = state_shape  # Input shape for the CNN
        self.action_size = action_size  # Number of actions (2: flap or do nothing)
        self.memory = deque(maxlen=2000)  # Experience replay memory
        self.gamma = 0.99  # Discount factor for future rewards
        self.epsilon = 1.0  # Initial exploration rate (100% random actions)
        self.epsilon_min = 0.01  # Minimum epsilon value
        self.epsilon_decay = 0.995  # Rate at which epsilon decays
        self.learning_rate = 0.00025  # Learning rate for optimizer
        self.model = self.build_model()  # Build the CNN model

    # Build the convolutional neural network for Q-learning
    def build_model(self):
        inputs = Input(shape=self.state_shape)  # Define input layer
        x = Conv2D(32, (8, 8), strides=4, activation='relu')(inputs)  # First conv layer
        x = Conv2D(64, (4, 4), strides=2, activation='relu')(x)  # Second conv layer
        x = Flatten()(x)  # Flatten conv output for dense layers
        x = Dense(128, activation='relu')(x)  # Fully connected hidden layer
        outputs = Dense(self.action_size, activation='linear')(x)  # Output layer for Q-values
        model = Model(inputs, outputs)  # Define the model
        model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=self.learning_rate), loss='mse')  # Compile with MSE loss
        return model

    # Store the experience in replay memory
    def remember(self, state, action, reward, next_state, done):
        self.memory.append((state, action, reward, next_state, done))

    # Decide what action to take based on current state
    def act(self, state):
        if np.random.rand() <= self.epsilon:
            return random.randrange(self.action_size)  # Choose random action
        q_values = self.model.predict(np.expand_dims(state, axis=0), verbose=0)  # Predict Q-values from model
        return np.argmax(q_values[0])  # Choose action with highest Q-value

    # Train the model using experiences from memory
    def replay(self, batch_size=32):
        if len(self.memory) < batch_size:
            return  # Not enough data to train

        minibatch = random.sample(self.memory, batch_size)  # Sample a random minibatch

        for state, action, reward, next_state, done in minibatch:
            target = self.model.predict(np.expand_dims(state, axis=0), verbose=0)[0]  # Predict current Q-values

            if done:
                target[action] = reward  # If game over, set Q-value to reward
            else:
                t = self.model.predict(np.expand_dims(next_state, axis=0), verbose=0)[0]  # Predict next state's Q-values
                target[action] = reward + self.gamma * np.amax(t)  # Bellman equation update

            self.model.fit(np.expand_dims(state, axis=0), np.expand_dims(target, axis=0), epochs=1, verbose=0)  # Train the model

        if self.epsilon > self.epsilon_min:
            self.epsilon *= self.epsilon_decay  # Decrease exploration over time

# Define input shape and number of actions, then initialize the agent
state_shape = (84, 84, 1)  # Shape of the preprocessed frame
action_size = len(actions)  # Typically 2: flap or do nothing
agent = DQNAgent(state_shape, action_size)  # Create the agent instance
```

```python
episodes = 20  # Number of training episodes to run
batch_size = 32  # Number of experiences to use in each training step

# Loop over each episode
for e in range(episodes):
    env.reset_game()  # Reset the game environment at the start of each episode
    frame = env.getScreenRGB()  # Capture the initial game screen
    state = preprocess_frame(frame)  # Preprocess the frame for model input

    total_reward = 0  # Initialize total reward counter
    done = False  # Track whether the game is over

    # Run the episode until the bird crashes
    while not env.game_over():
        action_idx = agent.act(state)  # Decide on an action (flap or do nothing)
        action = actions[action_idx]  # Translate index to actual action code
        reward = env.act(action)  # Perform the action in the game and get reward
        total_reward += reward  # Accumulate the reward

        next_frame = env.getScreenRGB()  # Capture the next screen frame
        next_state = preprocess_frame(next_frame)  # Preprocess the new frame

        # Store the experience for future training
        agent.remember(state, action_idx, reward, next_state, env.game_over())

        # Move to the next state
        state = next_state

        # Train the model using a batch from memory
        agent.replay(batch_size)

    # Output episode summary
    print(f"Episode {e+1}/{episodes} - Reward: {total_reward:.2f} - Epsilon: {agent.epsilon:.4f}")

    # Save the model every 5 episodes
    if (e + 1) % 5 == 0:
        agent.model.save(f"flappy_model_ep{e+1}.h5")
        print(f"✅ Model saved at episode {e+1}")
```

```python
from tensorflow.keras.models import load_model

# Load one of the saved models — pick ep5, ep10, ep15, or ep20
model = load_model("flappy_model_ep10.h5", compile=False)

# Reset the game environment
env.reset_game()
frame = env.getScreenRGB()
state = preprocess_frame(frame)
total_reward = 0

# Run a single test episode
while not env.game_over():
    q_values = model.predict(np.expand_dims(state, axis=0), verbose=0)
    action_idx = np.argmax(q_values[0])
    action = actions[action_idx]
    reward = env.act(action)
    total_reward += reward
    frame = env.getScreenRGB()
    state = preprocess_frame(frame)

print(f"Test Game Reward: {total_reward:.2f}")
```
